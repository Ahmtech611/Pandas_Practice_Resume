```python
# Now I move on to inter-pollation in pandas. So in this terminology, we actually put estimated (means frequent according to flow of data)
# value in missing values of dataset.

# So actually there are many types of Inter collation algorithms like linear interpollation and binomial interpollation , time series interpolation 
# and also other 8 to 9 types famous interpolation techniques or algorithms in Pandas.

# 1. Preserve Data Integrity.
# 2. Smooth trends(So actually its predicts arbitrary numbers like predictable numbers in missing value in dataset).
# 3. No loss of data.

# So in pandas we use interpolate() method for filling estimated values in missing data velues in dataset instead of deleting paticular missing
# vlaues containing rows or columns or putting default values. Its actually also good approach as compare to filling with calculated values like
# with mean() or mode() or medium specially where we need data thats actually predictable.

# Linear Interpolation :

# First lets create a dataset :

Data3 = {
    'Time' : [7, 12, None, 22, 27, None, 37, 42],
    'Value' : [12, 24, None, 48, None, 72, None, 96]
}

df3 = pd.DataFrame(Data3)

print(df3)
```

```python
# Now lets apply Linear interpolation :

df3['Value'] = df3['Value'].interpolate(method="linear")

print(df3)
```
