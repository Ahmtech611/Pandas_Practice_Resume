```python
# Now I talk about merging and sorting :

import pandas as pd
```

```python
df1 = pd.read_csv('DataSets&DataFiles/credit_card.csv')

df1
```

```python
df2 = pd.read_csv('DataSets&DataFiles/customer.csv')

df2

```

```python
df1.info()

```

```python
df2.info()
```

```python
# Now apply merge and join methods :

df_merged = pd.merge(df1, df2, on="Client_Num", how="outer")

print("Outer Join : ")
df_merged
```

```python
df_merged.info()
```

```python
# Inner Joining :

df_merged2 = pd.merge(df1, df2, on="Client_Num", how="inner")

print("Inner Joing : ")
df_merged2
```

```python
df_merged2.info()
```

```python
df_merged3 = pd.merge(df1, df2, on="Client_Num", how="left")

print("Left Joing : ")
df_merged3
```

```python
df_merged4 = pd.merge(df1, df2, on="Client_Num", how="right")

print("Right Joing : ")
df_merged4
```

```python
Sample_df1 = df1.head(5)
Sample_df2 = df2.head(5)

df_merged5 = pd.merge(Sample_df1, Sample_df2, how="cross")

print("Cross Joing : ")
df_merged5
```

```python
# Now the next topic is Concatenating the dataframes :

df_Customer1 = pd.read_excel('DataSets&DataFiles/df1_customers.xlsx')

df_Customer1
```

```python
df_Customer2 = pd.read_excel("DataSets&DataFiles/df2_customers.xlsx")

df_Customer2
```

```python
# Now we apply the concatenation :

# Vertical concatenation(Row Wise) :

df_Concat1 = pd.concat([df_Customer1, df_Customer2], axis=0, ignore_index=True)

df_Concat1
```

```python
# Horizontal concatenation(Column wise):

df_Concat2 = pd.concat([df_Customer1, df_Customer2], axis=1, ignore_index=True)

df_Concat2
```
