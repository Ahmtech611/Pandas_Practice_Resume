```python
df2 = pd.read_csv('DataSets&DataFiles/Bank_Accounts_Info.csv')

df2.info()
```

```python
df2.describe()
```

```python
# Inserting data at specific location or at the end(by default) :

# First insert at end position (by default) :

df2['Balance'] = [239800, 746000, 365366, 564700, 565360, 450000, 534900, 435000, 100, 637800]

print(df2)

```

```python
# Insert at specific position (index) column :

# Like Syntax : df2.insert(loc, 'Column_Name', data)

df2.insert(2, 'Monthly_Tax2', ((df2['Balance']/100)*2))

print(df2)
```

```python
df2.loc[8, 'Balance'] = 2500000

print(df2)
```

```python
df2.loc[8, 'Monthly_Tax2'] = 2500000 * 0.02

print(df2)
```

```python
# Updating multiple values :

df2['Balance'] = df2['Balance'] + ((df2['Balance']/100) * 5) # But Alternative Formula : df2['Balance'] = df2['Balance'] * 1.05

print(df2)
```

```python
# Then removing or drop the database :
df2.insert(2, 'Monthly_Tax2', [4796, 14920, 7307, 11294, 11307, 9000, 10698, 8700, 12700, 12756])

print(df2)
```

```python
# Then drop the column in dataset :

df2.drop(columns=['Monthly_Tax2'], inplace=True)

print(df2)
```

```python
df2.insert(4, 'Monthy_Tax', df2['Balance'] * 0.02)

print(df2)
```
