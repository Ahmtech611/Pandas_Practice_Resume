```python
# Now I work on GroupBy() functionality :

import pandas as pd
```

```python
data1 = {
    "Name" : ['Ahmad', 'Usman', 'Zainab', 'Fatima', 'Rehan'],
    'Age' : [23, 22, 23, 24, 26],
    'Salary' : [230000, 125000, 560000, 230000, 190000]
}

df = pd.DataFrame(data1)

df
```

```python
grouped = df.groupby('Age')['Salary'].sum()

print(grouped)
```

```python
# But If you have multiple columns then :

grouped2 = df.groupby(['Age', 'Name'])['Salary'].sum()

print(grouped2)
```
