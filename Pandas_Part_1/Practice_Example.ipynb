```python
# Select particular column by using square brackets.

# like df[column_name]

# if you want filtering your dataset based on dataset column in dataset :

# Filtering_dataset = df[df[column_name] > 3000] if we filter based one single condition.

# And if you want to filter based on multiple conditions like Filtering_dataset2 = df[df[column_name]>3000 & df[column_name2]< 200]

import pandas as pd

df = pd.read_csv('Data_sources&Data_sets/Employees_Info_dataset.csv')

df
```

```python
print(f'This column data represented as a seris : \n{df['Job Type']}')
```

```python
# Then now selecting multiple columns :

print(f"Now because of selecting multiple columns,\nthe data represented as a DataFrame : \n{df[['Company', 'Experience']]}")
```

```python
filtering_Employees_Info = df[df['Age']<24]

print(filtering_Employees_Info)
```

```python
# and now another example of filtering of rows in dataset in terms of multiple conditions :
# By using & operator filtering :

filtering_Employees_Info2 = df[(df['Age'] > 24) & (df['Salary(Rs)']< 1500000)]

print(filtering_Employees_Info2)
```

```python
# Now we filtered our dataset based onmultiple columns using | operator :

filtering_Employees_Info3 = df[(df['Age'] < 24) | (df['Performance Points']< 86)]

print(filtering_Employees_Info3)
```
