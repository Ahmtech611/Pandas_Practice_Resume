```python
# Also gcsfs library used to read data from paticular cloud storage like Google Cloud.

# You can also create new dataframes(data sets) by using dictionaries :

Data = {
    'Name' : ['Ahmad', 'Rehan', 'Faizan'],
    'Age' : [23, 24, 22],
    'Job' : ['Senior Data Scientist and Machine Learning Expert', 'Product Manager', 'Marketing Manager'],
    'Company' : ['Google', 'Microsoft', 'Nvidia']
}

df2 = pd.DataFrame(Data)

df2
```

```python
# Then convert this dataframe into CSV file by using to_csv() function :

df2.to_csv('Data_sources&Data_sets/Employees_data.csv', index=False)
df2
```

```python
# Then convert this dataFrame df2 into Excel format file (like xls or xlsx) :

df2.to_excel('Data_sources&Data_sets/Employees_dataset.xlsx', index=False)
df2
```

```python
# Then now convert this dataframe df2 into .JSON format.

df2.to_json('Data_sources&Data_sets/Employees_info.json')

df2
```
